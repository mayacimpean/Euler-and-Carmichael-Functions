##Introduction

This notebook will be used to write versions of the Euler totient and the Carmichael function. These functions will be used for testing examples and comparing running time.

##Function Implementation

```python
def prime_factorization(n):
  """Computes the prime factorization of a number n."""
  factors = {}
  d = 2
  temp = n
  while d * d <= temp:
    while temp % d == 0:
      factors[d] = factors.get(d, 0) + 1
      temp //= d
    d += 1
  if temp > 1:
    factors[temp] = factors.get(temp, 0) + 1
  return factors
```

```python
def carmichael_function(n):
  """Calculates the Carmichael function λ(n)."""
  if n == 1: return 1

  factors = prime_factorization(n)

  # Calculate lambda for each prime power factor
  lambdas = []
  for p, k in factors.items():
    lambdas.append(carmichael_lambda_prime_power(p, k))

  # Compute the LCM of all individual lambdas
  result = 1
  for val in lambdas:
    result = lcm(result, val)
  return result
```

## Euler Totient Function Implementation

```python
import time
import ipywidgets as widgets
from IPython.display import display, clear_output
import pandas as pd
import random # Import the random module

def euler_totient_function(n):
  """Calculates Euler's totient function φ(n)."""
  if n == 0: return 0
  if n == 1: return 1

  factors = prime_factorization(n)
  result = n
  for p in factors.keys():
    result -= result // p
  return result

# Global variable to store Euler Totient results
global_df_totient_results = pd.DataFrame()

# Demonstration with ipywidgets for a range
start_num_input_totient = widgets.Text(
    value='1',
    placeholder='Enter start number',
    description='Start Number:',
    disabled=False
)

end_num_input_totient = widgets.Text(
    value='10',
    placeholder='Enter end number',
    description='End Number:',
    disabled=False
)

num_runs_input_totient = widgets.Text(
    value='1',
    placeholder='Enter number of runs',
    description='Number of Runs:',
    disabled=False
)

button_totient = widgets.Button(description="Calculate Euler Totient for Range")
output_widget_totient = widgets.Output()

def on_button_totient_click(b):
    global global_df_totient_results # Declare intent to modify the global variable
    with output_widget_totient:
        clear_output()
        try:
            start_num = int(start_num_input_totient.value)
            end_num = int(end_num_input_totient.value)
            num_runs = int(num_runs_input_totient.value)

            if start_num > end_num:
                print("Start number cannot be greater than end number.")
                return
            if num_runs <= 0:
                print("Number of runs must be a positive integer.")
                return

            results = []
            # Create a list of numbers and shuffle it for random processing order
            numbers_to_process = list(range(start_num, end_num + 1))
            random.shuffle(numbers_to_process)

            for num in numbers_to_process:
                total_time = 0
                for _ in range(num_runs):
                    start_time = time.perf_counter()
                    totient_val = euler_totient_function(num)
                    end_time = time.perf_counter()
                    total_time += (end_time - start_time)

                average_time = total_time / num_runs
                results.append({'Number': num, 'Totient Value': totient_val, 'Average Time (s)': average_time})

            # Sort results by 'Number' before creating the DataFrame for consistent output
            results.sort(key=lambda x: x['Number'])
            global_df_totient_results = pd.DataFrame(results) # Assign to global variable
            display(global_df_totient_results)
        except ValueError:
            print("Please enter valid integers for the range and number of runs.")

button_totient.on_click(on_button_totient_click)

display(start_num_input_totient, end_num_input_totient, num_runs_input_totient, button_totient, output_widget_totient)
```

```python
import time
import ipywidgets as widgets
from IPython.display import display, clear_output
import pandas as pd
import random # Import the random module

# Global variable to store Carmichael function results
global_df_carmichael_results = pd.DataFrame()

def carmichael_lambda_prime_power(p, k):
  """Calculates λ(p^k) for a prime p and exponent k."""
  if p == 2:
    if k == 1: return 1
    if k == 2: return 2
    if k >= 3: return 2**(k-2)
  else: # p is an odd prime
    return p**(k-1) * (p-1)

def gcd(a, b):
  """Calculates the Greatest Common Divider of two numbers using the Euclidean algorithm."""
  while b:
    a, b = b, a % b
  return a

def lcm(a, b):
  """Calculates the Least Common Multiple of two numbers."""
  if a == 0 or b == 0: return 0
  return abs(a * b) // gcd(a, b)




# Demonstration with ipywidgets for Carmichael function over a range
start_num_input_carmichael = widgets.Text(
    value='1',
    placeholder='Enter start number',
    description='Start Number:',
    disabled=False
)

end_num_input_carmichael = widgets.Text(
    value='10',
    placeholder='Enter end number',
    description='End Number:',
    disabled=False
)

num_runs_input_carmichael = widgets.Text(
    value='1',
    placeholder='Enter number of runs',
    description='Number of Runs:',
    disabled=False
)

button_carmichael = widgets.Button(description="Calculate Carmichael Function for Range")
output_widget_carmichael = widgets.Output()

def on_button_carmichael_click(b):
    global global_df_carmichael_results # Declare intent to modify the global variable
    with output_widget_carmichael:
        clear_output()
        try:
            start_num = int(start_num_input_carmichael.value)
            end_num = int(end_num_input_carmichael.value)
            num_runs = int(num_runs_input_carmichael.value)

            if start_num > end_num:
                print("Start number cannot be greater than end number.")
                return
            if num_runs <= 0:
                print("Number of runs must be a positive integer.")
                return

            results = []
            # Create a list of numbers and shuffle it for random processing order
            numbers_to_process = list(range(start_num, end_num + 1))
            random.shuffle(numbers_to_process)

            for num in numbers_to_process:
                total_time = 0
                for _ in range(num_runs):
                    start_time = time.perf_counter()
                    carmichael_val = carmichael_function(num)
                    end_time = time.perf_counter()
                    total_time += (end_time - start_time)

                average_time = total_time / num_runs
                results.append({'Number': num, 'Carmichael Value': carmichael_val, 'Average Time (s)': average_time})

            # Sort results by 'Number' before creating the DataFrame for consistent output
            results.sort(key=lambda x: x['Number'])
            global_df_carmichael_results = pd.DataFrame(results) # Assign to global variable
            display(global_df_carmichael_results)
        except ValueError:
            print("Please enter valid integers for the range and number of runs.")

button_carmichael.on_click(on_button_carmichael_click)

display(start_num_input_carmichael, end_num_input_carmichael, num_runs_input_carmichael, button_carmichael, output_widget_carmichael)
```
